 Notebook 1: Data Loading & Cleaning

This notebook covers loading the IBM Telco dataset, handling missing values, type corrections, and preparing a clean dataset for analysis.

## 1.1 Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
pd.set_option('display.max_columns', None)
print("Libraries loaded ✓")
```

## 1.2 Load Dataset

```python
df = pd.read_csv('../data/raw/WA_FnUseC_TelcoCustomerChurn.csv')
print(f"Shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
df.head()
```

## 1.3 Initial Inspection

```python
print("=== DATA TYPES ===")
print(df.dtypes)
print("\n=== NULL VALUES ===")
print(df.isnull().sum())
print("\n=== BASIC STATS ===")
df.describe()
```

## 1.4 Handle Missing Values

The `TotalCharges` column contains 11 empty strings for customers with tenure=0 (new subscribers who haven't been billed yet). These are imputed with 0.0.

```python
# Check TotalCharges
print("Empty strings in TotalCharges:", (df['TotalCharges'] == '').sum())
print("Data type:", df['TotalCharges'].dtype)

# Fix: convert to numeric, coerce empties to NaN, fill with 0
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'].fillna(0.0, inplace=True)

print("\nAfter fix - nulls:", df['TotalCharges'].isnull().sum())
print("Dtype:", df['TotalCharges'].dtype)
```

## 1.5 Encode Target Variable

```python
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})
print("Churn distribution:")
print(df['Churn'].value_counts())
print(f"\nChurn rate: {df['Churn'].mean()*100:.2f}%")
```

## 1.6 Engineer Tenure Cohort

```python
bins   = [0, 12, 24, 48, 72]
labels = ['0-12 mo', '13-24 mo', '25-48 mo', '49-72 mo']
df['TenureCohort'] = pd.cut(df['tenure'], bins=bins, labels=labels)
print("Tenure cohort distribution:")
print(df['TenureCohort'].value_counts().sort_index())
```

## 1.7 Verify & Save

```python
print("=== POST-CLEANING QUALITY CHECK ===")
print(f"Total rows:      {len(df):,}")
print(f"Total columns:   {df.shape[1]}")
print(f"Null values:     {df.isnull().sum().sum()}")
print(f"Duplicates:      {df.duplicated().sum()}")
print(f"Churn rate:      {df['Churn'].mean()*100:.2f}%")

df.to_csv('../data/processed/cleaned_churn_data.csv', index=False)
print("\n✅ Cleaned dataset saved to data/processed/cleaned_churn_data.csv")
```
